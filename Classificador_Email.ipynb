# Resumo das métricas de classificação e seus significados:

**Precisão (Precision)**

    Definição: A proporção de verdadeiros positivos (TP) entre todos os exemplos que foram preditos como positivos (TP + FP).
    Significado: Indica a precisão das previsões positivas do modelo. Uma alta precisão significa poucos falsos positivos.

$\text{Precision} = \frac{TP}{TP + FP}$


**Recall (Sensibilidade)**

    Definição: A proporção de verdadeiros positivos (TP) entre todos os exemplos que são realmente positivos (TP + FN).
    Significado: Mede a capacidade do modelo de identificar corretamente as instâncias positivas. Um alto recall significa poucos falsos negativos.

$\text{Recall} = \frac{TP}{TP + FN}$

**F1-Score**

    Definição: A média harmônica da precisão e do recall (2 * (precision * recall) / (precision + recall)).
    Significado: Equilibra a precisão e o recall, sendo útil quando há um trade-off entre esses dois.

$
\text{F1-Score} = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}
$


**Acurácia (Accuracy)**

    Definição: A proporção de todas as previsões corretas (TP + TN) entre todas as instâncias (TP + TN + FP + FN).
    Significado: Mede a proporção de previsões corretas do modelo em relação a todas as instâncias.
$\text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN}$

**Macro Average**

    Definição: A média das métricas (precisão, recall, F1-score) calculada separadamente para cada classe e depois média das classes.
    Significado: Trata todas as classes igualmente, útil em datasets desbalanceados.
$\text{Macro Average} = \frac{1}{N} \sum_{i=1}^{N} \text{Métrica}_i$

**Weighted Average**

    Definição: A média ponderada das métricas (precisão, recall, F1-score) considerando o número de instâncias em cada classe.
    Significado: Dá mais peso às classes com mais instâncias, refletindo melhor o desempenho global do modelo em datasets desbalanceados.

$\text{Weighted Average} = \frac{1}{\sum_{i=1}^{N} |C_i|} \sum_{i=1}^{N} |C_i| \times \text{Métrica}_i$

Onde **N** é o número de classes e $|C_i|$é o número de instâncias na classe $i$.

Estas fórmulas são fundamentais para avaliar o desempenho de modelos de classificação, especialmente em datasets desbalanceados onde a escolha da métrica apropriada é crucial para obter uma avaliação precisa do modelo.

**Aplicações Práticas**

    Precisão: Útil quando o custo de um falso positivo é alto (e.g., classificação de emails importantes como spam).
    Recall: Crucial quando a detecção de todos os positivos é vital (e.g., diagnóstico de doenças).
    F1-Score: Ideal quando é necessário um balanço entre precisão e recall.
    Acurácia: Boa métrica geral, mas pode ser enganosa em datasets desbalanceados.
